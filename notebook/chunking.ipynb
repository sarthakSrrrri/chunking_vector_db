```python
# fetching all the data file inside data dir 

from pathlib import Path

DATA_DIR = Path("../data")

files = [
    file
    for file in DATA_DIR.rglob("*")
    if file.is_file()
    and file.suffix.lower() in {".pdf", ".txt", ".csv"}
]

print(f"Total files: {len(files)}")

for file in files:
    print(file)
```

```python
# st

documents = []

for file in files:
    documents.append({
        "source": file.name,
        "file_path": str(file),
        "file_type": file.suffix.lower(),
    })

print(f"Total documents: {len(documents)}")
print(documents)
```

```python
import pandas as pd
from pypdf import PdfReader


def load_document(document):
    file_path = Path(document["file_path"])
    file_type = document["file_type"]

    if file_type == ".txt":
        text = file_path.read_text(encoding="utf-8")

    elif file_type == ".csv":
        df = pd.read_csv(file_path)
        text = df.to_csv(index=False)

    elif file_type == ".pdf":
        reader = PdfReader(file_path)
        pages = []

        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                pages.append(page_text)

        text = "\n".join(pages)

    return {
        **document,
        "text": text
    }


loaded_documents = [load_document(doc) for doc in documents]

print(f"Loaded documents: {len(loaded_documents)}")
print(loaded_documents[0]["text"][:500])
```
